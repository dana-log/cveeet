```python
bike_df=bike_df.drop('end station id', axis=1)
bike_df=bike_df.drop('start station id', axis=1)
bike_df.info()
```

```python
bike_df['age']=2018-bike_df['birth year']
bike_df['age'].value_counts()
bike_df[bike_df['age']>60].info()
bike_df.drop('birth year', axis=1, inplace=True)
```

```python
bike_df['trip duration']=bike_df['stoptime'].dt.minute-bike_df['starttime'].dt.minute
```

```python
bike_df['starttime']=pd.to_datetime(bike_df['starttime'])
bike_df['starttime'].info()
bike_df['stoptime']=pd.to_datetime(bike_df['stoptime'])
bike_df['stoptime'].info()
bike_df['trip duration']=bike_df['stoptime'].dt.minute-bike_df['starttime'].dt.minute
bike_df.loc[3,'trip duration']
```

```python
bike_df['WeekdaySale'] = bike_df['starttime'].dt.dayofweek
weekend_count = bike_df[(bike_df['WeekdaySale'] == 5) | (bike_df['WeekdaySale'] == 6)].shape[0]
print(weekend_count)
```

```python
def daynight(starttime):
    if 0 <= starttime <= 6:
        return 'night'
    elif 6 < starttime <= 12:
        return 'morning'
    elif 12 < starttime <= 18:
        return 'day'
    return 'evening'
bike_df['time_of_day']=bike_df['starttime'].dt.hour.apply(daynight)
pred=bike_df['time_of_day'].value_counts()
answer=pred['day']/pred['night']
display(answer)
```
